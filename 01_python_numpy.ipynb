```python
import pandas as pd

data = {
    "name": ["Alice", "Bob", "Charlie"],
    "age": [25, 30, 35],
    "traffic_gbps": [2.5, 5.2, 3.8]
}

df = pd.DataFrame(data)

df
```

```python
import matplotlib.pyplot as plt

plt.plot(df["name"], df["traffic_gbps"])
plt.xlabel("Customer")
plt.ylabel("Traffic (Gbps)")
plt.title("Customer Traffic")

plt.show()
```
